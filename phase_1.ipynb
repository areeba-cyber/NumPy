Numpy array and basics

```python
import numpy as np
```

Creating array from list

```python
array_1d = np.array([1,2,3,4,5,6])
print("Array_1D :", array_1d)

array_2d = ([[1,2,3], [4,5,6]])
print("Array_2D :", array_2d)
# the array parameter requires a single array otherwise it gives a error
```

multiplication

```python
arr = np.array([1,2,3,4,5])
print(arr*2)
```

Python List VS Numpy array

```python
py_list = [1,2,3,4]
print("Python list multiplication:", py_list * 2)

np_array = np.array([1,2,3,4])
print("Numpy array multiplication:" , np_array*2)
```

Time Calculation

```python
import time
Start  = time.time()
py_list = [i*2 for i in range (1000000)]
print("\n List operation time:", time.time() - Start)

Start = time.time()
np_array = np.arange(1000000)*2
print("\n Array operation time:", time.time() - Start)
```

Creating array from scratch

```python
zeros = np.zeros((3,4))
print("Zeros Array:\n", zeros)

ones = np.ones((4,5))
print("Ones Array:\n", ones)

full = np.full((3,3),9)
print("Constant Array:\n", full)

random = np.random.random((4,5))
print("Random Array:\n", random)
# Random always fluctuates between 0s and 1s

sequence = np.arange(0,20,3)
# sequence = np.arange(start,stop,step)
print("Sequence Array:\n", sequence)
```

Vector / Tensor / Matrix

```python
vector = np.array([1,2,3])
print("Vector:", vector)

tensor = np.array([[[1,2],[3,4]],
                   [[5,6], [7,8]]])
print("Tensor:", tensor)

matrix = np.array([[1,2,3],
                   [4,5,6]])
print("Matrix:", matrix)
```

Array Properties

```python
array = np.array([[1,2,3],
                  [4,5,6]])
print("Shape:", array.shape)
print("Dimension:", array.ndim)
print("Size:", array.size)  
# size give the total number of elements
print("Data Type:", array.dtype)
```

Array Reshaping

```python
original = np.arange(12)
print("Original Array:", original)

reshaped = original.reshape((3,4))
print("\n Reshaped Array:", reshaped)

flattened = reshaped.flatten()
print("\n Flattened Array:", flattened)

raveled = reshaped.ravel()
print("\n Raveled Array:", raveled)
# Ravel returns view, instead of copy

Transpose = reshaped.T
print("\n Transposed Array:", Transpose)


```
