```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt 
```

```python
df = sns.load_dataset("tips")
df
```

```python
sns.countplot(x = df['sex'], hue=df['smoker'])
```

```python
# Bar Plot = y(quantitative) , x(categorical)
sns.barplot(x = df['sex'],y = df['total_bill'])
```
