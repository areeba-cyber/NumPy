```python
import matplotlib.pyplot as plt
import numpy as np
```

### Bar Chart(compare categories of data by representing each category with a bar)

```python
students = (["Ali", "Ayesha", "Hassan", "Ahmed"])
score = ([10, 20, 30, 40])

plt.title("Students Result", fontweight="bold")
plt.xlabel("Students", fontweight="bold")
plt.ylabel("Score", fontweight="bold")

plt.bar(students, score, color="pink")
# plt.barh(students, score, color="pink")
plt.show()
```

### Pie Chart (Circular chart divided into slices to show percentage of total / Good fpr visualizing distribution among categories)

```python
categories =["vegetables", "fruits", "fast-food", "drinks", "choclate"]
values = np.array([123, 200, 278, 435, 190])
colors = ["red", "pink", "purple", "yellow", "green"]
plt.pie(values, labels=categories,
                autopct="%1.1f%%",
                colors=colors,
                explode=[0.1,0,0,0,0],
                shadow=True,
                startangle=90
        )
plt.title("Monthly Expenses")
plt.show()
```

### Scatter Graph (Shows the relationship between two variables / Helps to identify a corelation [+,-,None] eg: Study hors VS result score)

```python
x = [0,1,5,2,6,2,8,9,7,5,3,0,5,4]
y = [10, 20, 49, 22, 56, 76, 88, 34, 48, 87, 56,38, 29, 62]
# alpha is transparency
x1 = [0,1,2,3,4,4,7,9,7,5,3,0,5,4]
y2 = [10, 20, 30, 40, 56, 46, 78, 50, 48, 87, 56,38, 29, 62]
plt.scatter(x,y, color="purple",
                #  alpha=0.5,
                 s = 100,
                 label="Class A")
plt.scatter(x1,y2, color="orange",
                #    alpha=0.5,
                   s = 100,
                   label="Class B")
plt.xlabel("Study Hours")
plt.ylabel("Reult")
plt.title("Results Calculation")
plt.legend()
plt.show()
```

### Histogram (A visual representation of the distribution of quantitative data / They group values into bins [intervals] and count how many fall in each range)

```python
scores = np.random.normal(loc=80, scale=10, size=100)
scores = np.clip(scores, 0, 100)

plt.hist(scores, bins=10,
                 edgecolor="blue")
plt.title("Exam Scores")
plt.xlabel("Results")
plt.ylabel("No. of Students")
plt.show()
```
