```python
import matplotlib.pyplot as plt
import numpy as np
```

### Bar Chart(compare categories of data by representing each category with a bar)

```python
students = (["Ali", "Ayesha", "Hassan", "Ahmed"])
score = ([10, 20, 30, 40])

plt.title("Students Result", fontweight="bold")
plt.xlabel("Students", fontweight="bold")
plt.ylabel("Score", fontweight="bold")

plt.bar(students, score, color="pink")
# plt.barh(students, score, color="pink")
plt.show()
```

### Pie Chart (Circular chart divided into slices to show percentage of total / Good fpr visualizing distribution among categories)

```python
categories =["vegetables", "fruits", "fast-food", "drinks", "choclate"]
values = np.array([123, 200, 278, 435, 190])
colors = ["red", "pink", "purple", "yellow", "green"]
plt.pie(values, labels=categories,
                autopct="%1.1f%%",
                colors=colors,
                explode=[0.1,0,0,0,0],
                shadow=True,
                startangle=90
        )
plt.title("Monthly Expenses")
plt.show()
```
