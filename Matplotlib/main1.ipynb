```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Basic
x = np.array([2021, 2022, 2023, 2024, 2025])
y1 = np.array([10, 30, 40, 29, 65])
y2 = np.array([8, 20, 50, 19, 30])
y3 = np.array([5, 14, 22, 38, 20])
line_style = dict(marker = ".",
                  markersize = 20,
                  markerfacecolor= "#1cd3fd",
                  markeredgecolor = "red",
                  linestyle = "solid",
                  linewidth = 3)
plt.plot(x, y1, color = "#1cfc45", **line_style)
plt.plot(x, y2, color = "purple", **line_style)
plt.plot(x, y3, color = "pink", **line_style)
plt.show()
```

### Labels

```python
x = np.array([2022, 2023, 2024, 2025, 2026])
y1 = np.array([15, 55, 30, 60, 35])
y2 = np.array([10, 20, 26, 40, 60])
y3 = np.array([11, 22, 40, 70, 45])

plt.title("Class Size", fontsize=20,
                        family="Arial",
                        fontweight = "bold",
                        color= "red")
plt.xlabel("Year",fontsize=20,
                  family="Arial",
                  fontweight = "bold",
                  color= "orange" )
plt.ylabel("Students", fontsize=20,
                       family="Arial",
                       fontweight = "bold",
                       color= "orange")
plt.tick_params(axis="both",
                colors="purple")

plt.plot(x,y1)
plt.plot(x,y2)
plt.plot(x,y3)
plt.xticks(x)

plt.show()
```

### Grid Lines(Hepls make plots easier to read by making reference lines)

```python
x = [1, 2, 3, 4]
y = [10, 30 , 20, 40]

plt.plot(x,y)
plt.grid(linewidth= 2,
         color="lightgray",
         linestyle="dashed")
# plt.grid(axis="x")
plt.show()
```
