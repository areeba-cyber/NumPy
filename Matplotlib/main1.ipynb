```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Basic
x = np.array([2021, 2022, 2023, 2024, 2025])
y1 = np.array([10, 30, 40, 29, 65])
y2 = np.array([8, 20, 50, 19, 30])
y3 = np.array([5, 14, 22, 38, 20])
line_style = dict(marker = ".",
                  markersize = 20,
                  markerfacecolor= "#1cd3fd",
                  markeredgecolor = "red",
                  linestyle = "solid",
                  linewidth = 3)
plt.plot(x, y1, color = "#1cfc45", **line_style)
plt.plot(x, y2, color = "purple", **line_style)
plt.plot(x, y3, color = "pink", **line_style)
plt.show()
```
