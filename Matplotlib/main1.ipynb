```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Basic
x = np.array([2021, 2022, 2023, 2024, 2025])
y1 = np.array([10, 30, 40, 29, 65])
y2 = np.array([8, 20, 10, 19, 30])
plt.plot(x, y1, marker = ".",
               markersize = 20,
               markerfacecolor= "#1cd3fd",
               markeredgecolor = "#1cd3fd",
               linestyle = "solid",
               linewidth = 3,
               color = "#1c5bfc")
plt.plot(x, y2, marker = ".",
               markersize = 20,
               markerfacecolor= "#1cd3fd",
               markeredgecolor = "#1cd3fd",
               linestyle = "solid",
               linewidth = 3,
               color = "#1c5bfc")
plt.show()
```
