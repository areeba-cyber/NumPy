Aggression [(Aggregate Functions = Reduces a set of values into single summary value), used to summarize and analyze data, used with groupby() function]

```python
import pandas as pd
```

```python
df = pd.read_csv("data.csv")
# Whole dataFrame
print(df.mean(numeric_only=True))
print(df.sum(numeric_only=True))
print(df.min(numeric_only=True))
print(df.max(numeric_only=True))
print(df.count())

# Single Column
print(df["age"].mean())
print(df["age"].min())
print(df["age"].max())
print(df["age"].sum())
print(df["age"].count())
```
