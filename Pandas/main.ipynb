```python
import pandas as pd
```

Series (A Pandas one dimensional labeled array that can work with any data type. Think of it as  a single column in a spreadsheet(1-D))


```python
data = [10,30,40,90]
data_2 = ["A","B","C"]
series = pd.Series(data)
series_2 = pd.Series(data_2)
# can be of any data type
print(series)
print(series_2)
```

change the indexing

```python
# indexing
data = [9,2,4,5]
series = pd.Series(data, index = ["a","b","c","d"])
print(series)

# directly access the value within a series
print(series.loc["b"])
print(series.loc["d"])
```

```python
# update the value in series
series.loc["c"] = 200
print(series)

# find value by integer
print(series.iloc[1])
```

Fitering

```python
data = [100,200,300,101,400,600]
series  = pd.Series(data, index = ["a", "b", "c", "d", "e", "f"])
print(series[series>=200])
```

```python
# Dictionary
calories = {
    "Day1" : 1200,
    "Day2" : 1600,
    "Day3" : 1700
}
series = pd.Series(calories)
print(series)
print(series.loc["Day1"])
series.loc["Day2"] = 1900
print(series)
print(series.iloc[0])
print(series[series > 1000])
```

DataFrame (A tabular data structure with rows and columns (2D))

```python
data = {
    "Name" : ["Ali", "Ahmed", "Areeba", "Amal", "Akbar"],
    "Age" : [34, 12, 56, 34, 18],
    "City" : ["UK", "KSA", "USA", "GERMANY", "PAK"]
}
df = pd.DataFrame(data)
df_2 = pd.DataFrame(data, index = ["Emp1", "Emp2", "Emp3", "Emp4", "Emp5"])
print(df)
print(df_2)
print(df_2.loc["Emp4"])
print(df_2.iloc[2])

# making a new column
df_2["Job"] = ["Cook", "Doc", "Engi", "Worker", "Chef"]
print(df_2)

# adding a new row
new_df = pd.DataFrame([{"Name":"Aga","Age":24,"City":"DUBAI","Job":"Watchman"}],index=["Emp5"])
df_2 = pd.concat([df_2, new_df])
print(df_2)
```
