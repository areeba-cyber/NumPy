Selection

```python
import pandas as pd
```

```python
df = pd.read_csv("data.csv")
# SELECTION BY COLUMN
print(df["name"])
# print(df["name"].to_string())
print(df[["name", "age"]])

```

```python
df = pd.read_csv("data.csv", index_col="name")
# SELECTION BY Row
print(df.loc["Hamza"])
print(df.loc["Hamza",["id","age"]])
print(df.loc["Hamza":"Hina",["id","age"]])
print(df.iloc[0:11])
# print(df.iloc[0:11:2])
print(df.iloc[0:11:2, 0:3])
# row, column
```

```python
df = pd.read_csv("data.csv", index_col="name")
student_name = input("Enter the student name:")
try:
    print(df.loc[student_name])
except KeyError:
    print(f"{student_name} not found")
```
