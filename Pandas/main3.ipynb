Filtering

```python
import pandas as pd
```

```python
df = pd.read_csv("data.csv")
# filtering = keeping the rows that match a condition
student_age = df[df["age"] >=22]
student_course = df[df["course"] == "Python"]
student_course_score = df[(df["course"] == "Python") | (df["score"] >= 70)]
print(student_age)
print(student_course)
print(student_course_score)
```
